## K1 — SQL Data Ingestion & Schema Validation

```python
# ╔══════════════════════════════════════════════════════════════════════════╗
# ║  K1 · DUCKDB INGESTION & SCHEMA VALIDATION                              ║
# ╚══════════════════════════════════════════════════════════════════════════╝
# Use ':memory:' for ephemeral analysis; swap to 'nigeria_fuel.duckdb'
# for a persistent warehouse you can query with DBeaver / dbt.
con = duckdb.connect(":memory:")

# ── Ingest CSV as native DuckDB table ────────────────────────────────────────
con.execute(f"""
    CREATE OR REPLACE TABLE raw_fuel_prices AS
    SELECT *
    FROM read_csv_auto(
        '{DATA_PATH}',
        header        = true,
        sample_size   = 10000,
        ignore_errors = true
    )
""")
print("✅  raw_fuel_prices table created.")

# ── SQL schema classification ─────────────────────────────────────────────────
schema_df = con.execute("""
    SELECT
        column_name,
        data_type,
        CASE
            WHEN column_name LIKE '%petrol_gasoline%'
             AND column_name NOT LIKE '%95_octane%'   THEN '✅ USABLE'
            WHEN column_name IN (
                 'ISO3','country','adm1_name','adm2_name','mkt_name',
                 'lat','lon','geo_id','DATES','year','month',
                 'currency','spatially_interpolated')  THEN '📋 METADATA'
            ELSE '🗑️  DROP (100 % NULL)'
        END AS classification
    FROM information_schema.columns
    WHERE table_name = 'raw_fuel_prices'
    ORDER BY classification, column_name
""").df()

# ── Plotly table ─────────────────────────────────────────────────────────────
cell_colors = schema_df["classification"].map({
    "✅ USABLE"          : PAL["green_lt"],
    "📋 METADATA"         : "#EEF4FF",
    "🗑️  DROP (100 % NULL)": PAL["red_lt"],
}).tolist()

fig_schema = go.Figure(go.Table(
    header=dict(
        values=["<b>Column</b>", "<b>DuckDB Type</b>", "<b>Classification</b>"],
        fill_color=PAL["blue"], font=dict(color="white", size=12),
        align="left", height=32,
    ),
    cells=dict(
        values=[schema_df[c].tolist() for c in ["column_name","data_type","classification"]],
        fill_color=["white", "white", cell_colors],
        align="left", height=26,
        font=dict(size=11),
    )
))
apply_theme(fig_schema, title="K1 · SQL Schema Audit — Column Classification", height=620)
fig_schema.show()
print(f"\nTotal columns: {len(schema_df)} | "
      f"Usable: {(schema_df.classification=='✅ USABLE').sum()} | "
      f"Drop: {(schema_df.classification=='🗑️  DROP (100 % NULL)').sum()}")
```

```python
# ── K1 · SQL Dataset Summary ─────────────────────────────────────────────────
summary = con.execute("""
    SELECT
        COUNT(*)                                          AS total_rows,
        COUNT(DISTINCT adm1_name)                         AS unique_states,
        COUNT(DISTINCT adm2_name)                         AS unique_lgas,
        COUNT(DISTINCT mkt_name)                          AS unique_markets,
        MIN(TRY_CAST(DATES AS DATE))::VARCHAR             AS earliest_date,
        MAX(TRY_CAST(DATES AS DATE))::VARCHAR             AS latest_date,
        COUNT(DISTINCT year)                              AS year_span,
        ROUND(AVG(c_fuel_petrol_gasoline),  2)::VARCHAR   AS mean_close_ngn,
        ROUND(MIN(c_fuel_petrol_gasoline),  2)::VARCHAR   AS min_close_ngn,
        ROUND(MAX(c_fuel_petrol_gasoline),  2)::VARCHAR   AS max_close_ngn
    FROM raw_fuel_prices
""").df()

# Transpose for display
s = summary.T.reset_index()
s.columns = ["Metric", "Value"]

fig_sum = go.Figure(go.Table(
    header=dict(
        values=["<b>Metric</b>", "<b>Value</b>"],
        fill_color=PAL["teal"], font=dict(color="white", size=12),
        align="left", height=32,
    ),
    cells=dict(
        values=[s["Metric"].tolist(), s["Value"].tolist()],
        align="left", height=28,
        font=dict(size=12),
        fill_color=[PAL["light"], "white"],
    )
))
apply_theme(fig_sum, title="K1 · Dataset Summary Statistics", height=380)
fig_sum.show()
```

## K2 — SQL Data Cleaning & Quality Audit

```python
# ╔══════════════════════════════════════════════════════════════════════════╗
# ║  K2 · SQL NULL AUDIT WITH PLOTLY HORIZONTAL BAR                         ║
# ╚══════════════════════════════════════════════════════════════════════════╝
null_audit = con.execute("""
    SELECT col, ROUND(100.0 * non_nulls / total, 2) AS coverage_pct
    FROM (
        SELECT 'o_fuel_petrol_gasoline'    AS col, COUNT(o_fuel_petrol_gasoline)    AS non_nulls, COUNT(*) AS total FROM raw_fuel_prices
        UNION ALL
        SELECT 'h_fuel_petrol_gasoline',        COUNT(h_fuel_petrol_gasoline),        COUNT(*) FROM raw_fuel_prices
        UNION ALL
        SELECT 'l_fuel_petrol_gasoline',        COUNT(l_fuel_petrol_gasoline),        COUNT(*) FROM raw_fuel_prices
        UNION ALL
        SELECT 'c_fuel_petrol_gasoline',        COUNT(c_fuel_petrol_gasoline),        COUNT(*) FROM raw_fuel_prices
        UNION ALL
        SELECT 'trust_fuel_petrol_gasoline',    COUNT(trust_fuel_petrol_gasoline),    COUNT(*) FROM raw_fuel_prices
        UNION ALL
        SELECT 'inflation_fuel_petrol_gasoline',COUNT(inflation_fuel_petrol_gasoline),COUNT(*) FROM raw_fuel_prices
        UNION ALL
        SELECT 'fuel_petrol_gasoline (raw)',    COUNT(fuel_petrol_gasoline),          COUNT(*) FROM raw_fuel_prices
        UNION ALL
        SELECT 'fuel_diesel',                  COUNT(fuel_diesel),                   COUNT(*) FROM raw_fuel_prices
        UNION ALL
        SELECT 'fuel_gas',                     COUNT(fuel_gas),                      COUNT(*) FROM raw_fuel_prices
        UNION ALL
        SELECT 'fuel_kerosene',                COUNT(fuel_kerosene),                 COUNT(*) FROM raw_fuel_prices
        UNION ALL
        SELECT 'fuel_petrol_95_octane',        COUNT(fuel_petrol_gasoline_95_octane),COUNT(*) FROM raw_fuel_prices
        UNION ALL
        SELECT 'fuel_super_petrol',            COUNT(fuel_super_petrol),             COUNT(*) FROM raw_fuel_prices
    ) t
    ORDER BY coverage_pct DESC
""").df()

# Colour by coverage tier
def bar_color(v):
    if v == 100:   return PAL["teal"]
    if v >= 50:    return PAL["amber"]
    if v > 0:      return PAL["coral"]
    return "#D0D0D0"

null_audit["color"] = null_audit["coverage_pct"].apply(bar_color)

fig_null = go.Figure(go.Bar(
    x=null_audit["coverage_pct"],
    y=null_audit["col"],
    orientation="h",
    marker_color=null_audit["color"],
    text=null_audit["coverage_pct"].apply(lambda v: f"{v}%"),
    textposition="outside",
    hovertemplate="<b>%{y}</b><br>Coverage: %{x}%<extra></extra>",
))
fig_null.add_vline(x=100, line_dash="dot", line_color=PAL["slate"], line_width=1.5)
apply_theme(fig_null, title="K2 · Column Data Coverage Audit (SQL NULL Analysis)",
            height=440, x_title="Coverage (%)", y_title="Column")
fig_null.update_xaxes(range=[0, 115])
fig_null.show()

print("\n📌 Decision: drop all 0%-coverage columns.")
print("   Retain: o/h/l/c_fuel_petrol_gasoline, trust_, inflation_, lat, lon, geo columns.")
```

```python
# ╔══════════════════════════════════════════════════════════════════════════╗
# ║  K2 · CREATE CLEAN ANALYTICAL TABLE (ALL SQL)                           ║
# ╚══════════════════════════════════════════════════════════════════════════╝
# Design decisions encoded in SQL (documented for thesis methodology chapter):
#   1. Cast DATES → DATE type for window functions
#   2. Regime flag: subsidy removed 29 May 2023 (Tinubu inauguration)
#   3. Exclude synthetic 'Market Average' aggregate for LGA-level work
#   4. Retain is_national_avg flag to allow fast national-level queries

con.execute("""
    CREATE OR REPLACE TABLE clean_fuel_prices AS
    SELECT
        -- ── Geography ──────────────────────────────────────────────────────
        ISO3,
        country,
        adm1_name,
        adm2_name,
        mkt_name,
        geo_id,
        lat,
        lon,

        -- ── Time ───────────────────────────────────────────────────────────
        TRY_CAST(DATES AS DATE)                          AS price_date,
        year,
        month,
        currency,
        spatially_interpolated,

        -- ── OHLC Price Series (₦ / litre) ──────────────────────────────────
        o_fuel_petrol_gasoline                           AS price_open,
        h_fuel_petrol_gasoline                           AS price_high,
        l_fuel_petrol_gasoline                           AS price_low,
        c_fuel_petrol_gasoline                           AS price_close,

        -- ── Data Quality Weight ────────────────────────────────────────────
        trust_fuel_petrol_gasoline                       AS trust_score,

        -- ── Inflation Signal ───────────────────────────────────────────────
        inflation_fuel_petrol_gasoline                   AS fuel_inflation_mom,

        -- ── Policy Regime Flag ─────────────────────────────────────────────
        -- 0 = subsidised era  |  1 = post-subsidy / market-reflective
        CASE
            WHEN year < 2023                      THEN 0
            WHEN year = 2023 AND month < 6        THEN 0
            ELSE 1
        END                                              AS post_subsidy_regime,

        -- ── Aggregate Row Flag ─────────────────────────────────────────────
        CASE WHEN adm1_name = 'Market Average'    THEN 1
             ELSE 0
        END                                              AS is_national_avg

    FROM raw_fuel_prices
    WHERE c_fuel_petrol_gasoline IS NOT NULL
    ORDER BY geo_id, TRY_CAST(DATES AS DATE)
""")

n = con.execute("SELECT COUNT(*) FROM clean_fuel_prices").fetchone()[0]
print(f"✅  clean_fuel_prices → {n:,} rows")
```

```python
# ╔══════════════════════════════════════════════════════════════════════════╗
# ║  K2 · PANDAS INTERPOLATION FOR INFLATION NULLs (5.2%)                  ║
# ║      DuckDB does not support per-group linear interpolation natively.    ║
# ║      We pull → interpolate → push back as a SQL-registered view.        ║
# ╚══════════════════════════════════════════════════════════════════════════╝
df = con.execute("SELECT * FROM clean_fuel_prices").df()
df["price_date"] = pd.to_datetime(df["price_date"])

before_nulls = df["fuel_inflation_mom"].isna().sum()

df["fuel_inflation_mom"] = (
    df.sort_values("price_date")
      .groupby("geo_id")["fuel_inflation_mom"]
      .transform(lambda s: s.interpolate(method="linear", limit_direction="both"))
)

after_nulls = df["fuel_inflation_mom"].isna().sum()

# Push interpolated frame back into DuckDB
con.register("df_interpolated", df)
con.execute("CREATE OR REPLACE TABLE clean_fuel_prices AS SELECT * FROM df_interpolated")

print(f"✅  Interpolation: inflation NULLs {before_nulls:,} → {after_nulls}")
print(f"   Shape: {df.shape}")
df[["price_date","adm1_name","price_open","price_high","price_low","price_close",
    "trust_score","post_subsidy_regime"]].head(4)
```

## K3 — SQL Feature Engineering

```python
# ╔══════════════════════════════════════════════════════════════════════════╗
# ║  K3 · SQL WINDOW-FUNCTION FEATURE ENGINEERING                           ║
# ║      All derived features computed in a single SQL CTE pass.            ║
# ║      This mirrors a production dbt model.                               ║
# ╚══════════════════════════════════════════════════════════════════════════╝
con.execute("""
    CREATE OR REPLACE TABLE engineered_features AS
    WITH base AS (
        SELECT * FROM clean_fuel_prices
    )
    SELECT
        base.*,

        -- ── OHLC Derivatives ─────────────────────────────────────────────
        -- Intra-month price range (volatility proxy)
        (price_high - price_low)                                 AS spread,

        -- Within-month directional momentum
        (price_close - price_open)                               AS momentum,

        -- Normalised volatility (% of open)
        ROUND(100.0*(price_high - price_low)/NULLIF(price_open,0), 4)
                                                                 AS spread_pct,

        -- ── Lag Features (per market geo_id) ─────────────────────────────
        LAG(price_close, 1)  OVER w                              AS lag1_close,
        LAG(price_close, 3)  OVER w                              AS lag3_close,
        LAG(price_close, 6)  OVER w                              AS lag6_close,
        LAG(price_close, 12) OVER w                              AS lag12_close,

        -- ── Rolling Statistics ────────────────────────────────────────────
        ROUND(AVG(price_close)    OVER (PARTITION BY geo_id
              ORDER BY price_date ROWS BETWEEN  2 PRECEDING AND CURRENT ROW), 2)
                                                                 AS roll_mean_3m,
        ROUND(AVG(price_close)    OVER (PARTITION BY geo_id
              ORDER BY price_date ROWS BETWEEN 11 PRECEDING AND CURRENT ROW), 2)
                                                                 AS roll_mean_12m,
        ROUND(STDDEV(price_close) OVER (PARTITION BY geo_id
              ORDER BY price_date ROWS BETWEEN 11 PRECEDING AND CURRENT ROW), 2)
                                                                 AS roll_std_12m,

        -- ── MoM & YoY % Change ────────────────────────────────────────────
        ROUND(
            100.0*(price_close - LAG(price_close,1) OVER w)
            / NULLIF(LAG(price_close,1) OVER w, 0), 4)          AS mom_pct,
        ROUND(
            100.0*(price_close - LAG(price_close,12) OVER w)
            / NULLIF(LAG(price_close,12) OVER w, 0), 4)         AS yoy_pct,

        -- ── Trust-Weighted Price ─────────────────────────────────────────
        price_close * trust_score                                AS trust_wtd_close,

        -- ── Cyclical Month Encoding (prevents 1 & 12 numerical gap) ──────
        SIN(2 * 3.14159265 * month / 12.0)                      AS month_sin,
        COS(2 * 3.14159265 * month / 12.0)                      AS month_cos

    FROM base
    WINDOW w AS (PARTITION BY geo_id ORDER BY price_date)
    ORDER BY geo_id, price_date
""")

feat_df = con.execute("SELECT * FROM engineered_features").df()
feat_df["price_date"] = pd.to_datetime(feat_df["price_date"])

print(f"✅  engineered_features → {feat_df.shape[0]:,} rows × {feat_df.shape[1]} columns")

# ── Plotly table preview ──────────────────────────────────────────────────────
preview_cols = ["price_date","adm1_name","price_close","spread",
                "momentum","roll_mean_3m","mom_pct","trust_score",
                "month_sin","month_cos","post_subsidy_regime"]
prev = feat_df[preview_cols].head(8).round(3)

fig_prev = go.Figure(go.Table(
    header=dict(
        values=[f"<b>{c}</b>" for c in prev.columns],
        fill_color=PAL["purple"], font=dict(color="white", size=11),
        align="left", height=30,
    ),
    cells=dict(
        values=[prev[c].tolist() for c in prev.columns],
        align="left", height=26,
        font=dict(size=10),
        fill_color=[[PAL["light"] if i%2==0 else "white" for i in range(len(prev))]]*len(prev.columns),
    )
))
apply_theme(fig_prev, title="K3 · Engineered Features Preview (first 8 rows)", height=340)
fig_prev.show()
```

## K4 — Insight 1: Asymmetric Duopoly Shift (2024–2026)
Quantify the market-structure change from NNPCL monopoly to NNPCL + Dangote duopoly using OHLC candlesticks and cross-market dispersion.

```python
# ╔══════════════════════════════════════════════════════════════════════════╗
# ║  K4 · ASYMMETRIC DUOPOLY — CANDLESTICK + DISPERSION                     ║
# ╚══════════════════════════════════════════════════════════════════════════╝
duopoly_df = con.execute("""
    SELECT
        price_date,
        post_subsidy_regime,
        ROUND(AVG(price_open),   2) AS avg_open,
        ROUND(AVG(price_high),   2) AS avg_high,
        ROUND(AVG(price_low),    2) AS avg_low,
        ROUND(AVG(price_close),  2) AS avg_close,
        ROUND(STDDEV(price_close),2) AS cross_mkt_std,
        ROUND(AVG(mom_pct),      4) AS avg_mom_pct,
        ROUND(AVG(spread),       2) AS avg_spread
    FROM engineered_features
    WHERE is_national_avg = 0
    GROUP BY price_date, post_subsidy_regime
    ORDER BY price_date
""").df()
duopoly_df["price_date"] = pd.to_datetime(duopoly_df["price_date"])

# ── Regime comparison stats table ────────────────────────────────────────────
regime_stats = con.execute("""
    SELECT
        CASE post_subsidy_regime
            WHEN 0 THEN '0 — Subsidised (2007 – May 2023)'
            ELSE        '1 — Post-Subsidy / Duopoly (Jun 2023 – 2026)'
        END                                     AS regime,
        ROUND(AVG(price_close),    2)            AS mean_price_ngn,
        ROUND(STDDEV(price_close), 2)            AS std_dev_ngn,
        ROUND(MIN(price_close),    2)            AS min_price_ngn,
        ROUND(MAX(price_close),    2)            AS max_price_ngn,
        ROUND(AVG(mom_pct),        3)            AS avg_mom_pct,
        ROUND(AVG(spread),         2)            AS avg_intramonth_spread
    FROM engineered_features
    WHERE is_national_avg = 0
    GROUP BY post_subsidy_regime
    ORDER BY post_subsidy_regime
""").df()

# ── Figure: 3-panel (candlestick | dispersion | MoM change) ─────────────────
fig_duopoly = make_subplots(
    rows=3, cols=1,
    shared_xaxes=True,
    row_heights=[0.50, 0.25, 0.25],
    subplot_titles=(
        "National Average OHLC Candlestick (₦/L)",
        "Cross-Market Price Dispersion (Std Dev)",
        "Month-on-Month % Change",
    ),
    vertical_spacing=0.06,
)

# Panel 1 — Candlestick
fig_duopoly.add_trace(go.Candlestick(
    x=duopoly_df["price_date"],
    open=duopoly_df["avg_open"],
    high=duopoly_df["avg_high"],
    low=duopoly_df["avg_low"],
    close=duopoly_df["avg_close"],
    increasing_line_color=PAL["teal"],
    decreasing_line_color=PAL["coral"],
    name="OHLC",
), row=1, col=1)

# Regime shading — subsidy removal
for fig_r in [fig_duopoly]:
    fig_r.add_vrect(
        x0="2023-05-01", x1="2023-09-01",
        fillcolor=PAL["amber"], opacity=0.18,
        line_width=0, annotation_text="Subsidy removal",
        annotation_position="top left",
        annotation_font_size=10,
        row=1, col=1,
    )
    fig_r.add_vrect(
        x0="2024-01-01", x1=str(duopoly_df["price_date"].max().date()),
        fillcolor=PAL["purple"], opacity=0.07,
        line_width=0, annotation_text="Asymmetric duopoly era",
        annotation_position="top left",
        annotation_font_size=10,
        row=1, col=1,
    )

# Panel 2 — Cross-market Std Dev bar
fig_duopoly.add_trace(go.Bar(
    x=duopoly_df["price_date"],
    y=duopoly_df["cross_mkt_std"],
    marker_color=[
        PAL["coral"] if r == 1 else PAL["slate"]
        for r in duopoly_df["post_subsidy_regime"]
    ],
    name="Cross-market Std Dev",
    hovertemplate="%{x|%b %Y}<br>Std Dev: ₦%{y:.2f}<extra></extra>",
), row=2, col=1)

# Panel 3 — MoM % change line
fig_duopoly.add_trace(go.Scatter(
    x=duopoly_df["price_date"],
    y=duopoly_df["avg_mom_pct"],
    mode="lines",
    line=dict(color=PAL["blue"], width=1.8),
    fill="tozeroy",
    fillcolor="rgba(45,125,210,0.12)",
    name="MoM %",
    hovertemplate="%{x|%b %Y}<br>MoM: %{y:.2f}%<extra></extra>",
), row=3, col=1)
fig_duopoly.add_hline(y=0, line_dash="dot", line_color=PAL["slate"],
                      line_width=1, row=3, col=1)

fig_duopoly.update_layout(
    title=dict(text="K4 · Market Structure Shift: NNPCL Monopoly → Asymmetric Duopoly",
               x=0.04, font=dict(size=16, color="#1A1A2E")),
    height=720,
    template="plotly_white",
    paper_bgcolor="white",
    plot_bgcolor="#F8F9FA",
    showlegend=False,
    xaxis_rangeslider_visible=False,
    font=dict(family="Inter, Arial, sans-serif", size=12),
    margin=dict(l=60, r=30, t=70, b=40),
)
fig_duopoly.update_yaxes(showgrid=True, gridcolor="#ECECEC",
                          tickprefix="₦", row=1, col=1)
fig_duopoly.update_yaxes(showgrid=True, gridcolor="#ECECEC",
                          tickprefix="₦", row=2, col=1)
fig_duopoly.update_yaxes(showgrid=True, gridcolor="#ECECEC",
                          ticksuffix="%", row=3, col=1)
fig_duopoly.show()

# Regime comparison table
fig_rs = go.Figure(go.Table(
    header=dict(
        values=[f"<b>{c}</b>" for c in regime_stats.columns],
        fill_color=PAL["blue"], font=dict(color="white", size=11),
        align="left", height=32,
    ),
    cells=dict(
        values=[regime_stats[c].tolist() for c in regime_stats.columns],
        fill_color=[[PAL["green_lt"],"#FFF4E0"]]*len(regime_stats.columns),
        align="left", height=28, font=dict(size=11),
    )
))
apply_theme(fig_rs, title="K4 · Regime Comparison — Key Statistics", height=200)
fig_rs.show()
```

## K5 — Insight 2: Volatility Squeeze on MSMEs
Flag months where fuel price surges breach a 5% MoM threshold and correlate with a PMI proxy.

```python
# ╔══════════════════════════════════════════════════════════════════════════╗
# ║  K5 · MSME VOLATILITY SQUEEZE — PLOTLY MULTI-PANEL                      ║
# ╚══════════════════════════════════════════════════════════════════════════╝
vol_df = con.execute("""
    WITH monthly AS (
        SELECT
            price_date,
            post_subsidy_regime,
            ROUND(AVG(price_close),   2) AS avg_close,
            ROUND(AVG(mom_pct),       4) AS avg_mom,
            ROUND(AVG(spread_pct),    4) AS avg_spread_pct,
            ROUND(STDDEV(price_close),2) AS cross_std
        FROM engineered_features
        WHERE is_national_avg = 0
        GROUP BY price_date, post_subsidy_regime
    ),
    with_roll AS (
        SELECT *,
            CASE WHEN avg_mom > 5 THEN 1 ELSE 0 END                      AS is_spike,
            AVG(avg_close) OVER (
                ORDER BY price_date
                ROWS BETWEEN 5 PRECEDING AND CURRENT ROW)                AS roll6_mean,
            STDDEV(avg_close) OVER (
                ORDER BY price_date
                ROWS BETWEEN 5 PRECEDING AND CURRENT ROW)                AS roll6_std
        FROM monthly
    )
    SELECT * FROM with_roll ORDER BY price_date
""").df()
vol_df["price_date"] = pd.to_datetime(vol_df["price_date"])

spikes = vol_df[vol_df["is_spike"] == 1]

# ── Synthetic PMI (replace with Stanbic IBTC CSV) ───────────────────────────
pmi_start = vol_df[vol_df["price_date"] >= "2020-01-01"]
pmi_dates  = pd.date_range("2020-01", periods=len(pmi_start), freq="MS")
pmi_vals   = (52 - pmi_start["avg_mom"].values * 0.9
              + np.random.normal(0, 1.2, len(pmi_start)))
pmi_df_k5  = pd.DataFrame({"price_date": pmi_dates, "pmi": pmi_vals.clip(40, 58)})

# ── 3-panel Plotly figure ────────────────────────────────────────────────────
fig_vol = make_subplots(
    rows=3, cols=1,
    shared_xaxes=False,
    row_heights=[0.40, 0.30, 0.30],
    subplot_titles=(
        "MoM Price Change — Spike Detection (MSME Alert Threshold = 5%)",
        "6-Month Rolling Volatility (₦ Std Dev) — MSME Planning Horizon",
        "PMI Index vs Fuel Spike Events (2020–2026) — Replace w/ Stanbic IBTC CSV",
    ),
    vertical_spacing=0.10,
)

# Panel 1 — MoM change area + spike markers
fig_vol.add_trace(go.Scatter(
    x=vol_df["price_date"], y=vol_df["avg_mom"],
    mode="lines", name="MoM %",
    line=dict(color=PAL["blue"], width=1.8),
    fill="tozeroy", fillcolor="rgba(45,125,210,0.10)",
    hovertemplate="%{x|%b %Y}<br>MoM: %{y:.2f}%<extra></extra>",
), row=1, col=1)
fig_vol.add_trace(go.Scatter(
    x=spikes["price_date"], y=spikes["avg_mom"],
    mode="markers", name="⚠ Spike (>5%)",
    marker=dict(color=PAL["coral"], size=10, symbol="triangle-up",
                line=dict(color="white", width=1)),
    hovertemplate="<b>SPIKE</b><br>%{x|%b %Y}<br>MoM: %{y:.2f}%<extra></extra>",
), row=1, col=1)
fig_vol.add_hline(y=5,  line_dash="dot", line_color=PAL["coral"],
                  line_width=1.5, annotation_text="5% alert",
                  annotation_position="right", row=1, col=1)
fig_vol.add_hline(y=0,  line_dash="dot", line_color=PAL["slate"],
                  line_width=1, row=1, col=1)

# Panel 2 — Rolling 6-month std dev
fig_vol.add_trace(go.Scatter(
    x=vol_df["price_date"], y=vol_df["roll6_std"].fillna(0),
    mode="lines", name="6m Std Dev",
    line=dict(color=PAL["amber"], width=0),
    fill="tozeroy", fillcolor=PAL["amber"],
    opacity=0.65,
    hovertemplate="%{x|%b %Y}<br>6m σ: ₦%{y:.2f}<extra></extra>",
), row=2, col=1)

# Panel 3 — PMI
fig_vol.add_trace(go.Scatter(
    x=pmi_df_k5["price_date"], y=pmi_df_k5["pmi"],
    mode="lines+markers", name="PMI (proxy)",
    line=dict(color=PAL["purple"], width=2),
    marker=dict(size=4),
    hovertemplate="%{x|%b %Y}<br>PMI: %{y:.1f}<extra></extra>",
), row=3, col=1)
fig_vol.add_hline(y=50, line_dash="dash", line_color=PAL["slate"],
                  line_width=1.5, annotation_text="Expansion/Contraction",
                  annotation_position="right", row=3, col=1)

# Mark spike months on PMI panel too
spike_pmi = pmi_df_k5[pmi_df_k5["price_date"].isin(spikes["price_date"].values)]
fig_vol.add_trace(go.Scatter(
    x=spike_pmi["price_date"], y=spike_pmi["pmi"],
    mode="markers", name="Fuel spike overlap",
    marker=dict(color=PAL["coral"], size=9, symbol="x"),
), row=3, col=1)

fig_vol.update_layout(
    title=dict(text="K5 · MSME Volatility Squeeze — Cost-Push Inflation Analysis",
               x=0.04, font=dict(size=16, color="#1A1A2E")),
    height=780, template="plotly_white", paper_bgcolor="white",
    plot_bgcolor="#F8F9FA",
    font=dict(family="Inter, Arial, sans-serif", size=12),
    margin=dict(l=60, r=30, t=80, b=40),
    legend=dict(bgcolor="rgba(255,255,255,0.85)", bordercolor="#DDD", borderwidth=1),
)
fig_vol.update_yaxes(showgrid=True, gridcolor="#ECECEC",
                      ticksuffix="%", row=1, col=1)
fig_vol.update_yaxes(showgrid=True, gridcolor="#ECECEC",
                      tickprefix="₦", row=2, col=1)
fig_vol.update_yaxes(showgrid=True, gridcolor="#ECECEC", row=3, col=1)
fig_vol.show()

n_spikes = int(vol_df["is_spike"].sum())
print(f"\n⚠️   Spike months (MoM > 5%): {n_spikes} of {len(vol_df)} "
      f"({100*n_spikes/len(vol_df):.1f}% of the series)")
```

## K6 — Insight 3: Regional Price Disparity
Map geospatial price gaps between states and geopolitical zones. Targets for cash-transfer programmes.

```python
# ╔══════════════════════════════════════════════════════════════════════════╗
# ║  K6 · REGIONAL PRICE DISPARITY — PLOTLY BAR + TREEMAP                   ║
# ╚══════════════════════════════════════════════════════════════════════════╝
regional_df = con.execute("""
    WITH state_agg AS (
        SELECT
            adm1_name                           AS state,
            post_subsidy_regime,
            ROUND(AVG(price_close), 2)          AS avg_price,
            ROUND(STDDEV(price_close), 2)       AS price_std,
            ROUND(AVG(trust_score), 4)          AS avg_trust,
            COUNT(*)                            AS obs
        FROM engineered_features
        WHERE is_national_avg = 0
          AND adm1_name NOT IN ('Market Average')
        GROUP BY adm1_name, post_subsidy_regime
    ),
    nat AS (
        SELECT post_subsidy_regime,
               AVG(avg_price) AS nat_mean
        FROM state_agg GROUP BY post_subsidy_regime
    )
    SELECT
        s.*,
        ROUND(n.nat_mean, 2)                           AS nat_mean,
        ROUND(s.avg_price - n.nat_mean, 2)             AS premium_ngn,
        ROUND(100.0*(s.avg_price - n.nat_mean)
              / NULLIF(n.nat_mean, 0), 2)              AS premium_pct
    FROM state_agg s
    JOIN nat n USING (post_subsidy_regime)
    ORDER BY post_subsidy_regime DESC, avg_price DESC
""").df()

post_sub = regional_df[regional_df["post_subsidy_regime"]==1].copy()
post_sub = post_sub.sort_values("avg_price", ascending=True)

# ── Panel A: Horizontal bar — avg price by state ─────────────────────────────
fig_bar = go.Figure(go.Bar(
    x=post_sub["avg_price"],
    y=post_sub["state"],
    orientation="h",
    marker=dict(
        color=post_sub["premium_pct"],
        colorscale=[[0, PAL["teal"]], [0.5, "#FFFFFF"], [1, PAL["coral"]]],
        colorbar=dict(title="Premium %", thickness=14),
        cmin=post_sub["premium_pct"].min(),
        cmid=0,
        cmax=post_sub["premium_pct"].max(),
    ),
    text=post_sub["avg_price"].apply(lambda v: f"₦{v:,.0f}"),
    textposition="outside",
    hovertemplate="<b>%{y}</b><br>Avg Price: ₦%{x:,.2f}<br>Premium: %{customdata:.1f}%<extra></extra>",
    customdata=post_sub["premium_pct"],
))
nat_mean_val = float(post_sub["nat_mean"].iloc[0])
fig_bar.add_vline(
    x=nat_mean_val,
    line_dash="dash", line_color=PAL["slate"], line_width=2,
    annotation_text=f"National avg: ₦{nat_mean_val:,.0f}",
    annotation_position="top right",
    annotation_font_size=11,
)
apply_theme(
    fig_bar,
    title="K6 · State Retail Price Ranking — Post-Subsidy Era (Jun 2023–2026)",
    height=max(500, len(post_sub)*22),
    x_title="Avg Retail Price (₦/L)",
)
fig_bar.show()

# ── Panel B: Treemap — price premium/discount ────────────────────────────────
post_sub["abs_premium"] = post_sub["premium_ngn"].abs().clip(lower=1)
post_sub["label"] = post_sub.apply(
    lambda r: f"{r['state']}<br>₦{r['avg_price']:,.0f}<br>{r['premium_pct']:+.1f}%", axis=1
)

fig_tree = px.treemap(
    post_sub,
    path=["state"],
    values="abs_premium",
    color="premium_pct",
    color_continuous_scale=[[0,PAL["teal"]],[0.5,"#FFFDE4"],[1,PAL["coral"]]],
    color_continuous_midpoint=0,
    custom_data=["state","avg_price","premium_pct"],
    title="K6 · Price Premium Treemap — States Above/Below National Average",
)
fig_tree.update_traces(
    hovertemplate="<b>%{customdata[0]}</b><br>"
                  "Avg Price: ₦%{customdata[1]:,.2f}<br>"
                  "Premium: %{customdata[2]:+.1f}%<extra></extra>",
    texttemplate="<b>%{label}</b><br>₦%{customdata[1]:,.0f}<br>%{customdata[2]:+.1f}%",
)
apply_theme(fig_tree, height=520)
fig_tree.show()

top5 = post_sub.sort_values("avg_price", ascending=False).head(5)
bot5 = post_sub.sort_values("avg_price", ascending=True).head(5)
print("\n📊 5 most expensive states:",
      list(zip(top5["state"], top5["avg_price"])))
print("📊 5 cheapest states:      ",
      list(zip(bot5["state"], bot5["avg_price"])))
```

## K7 — Insight 4: Structural Break Detection (PELT Algorithm)
Replace subjective 2023 assumption with objective, data-driven change-point dates.

```python
# ╔══════════════════════════════════════════════════════════════════════════╗
# ║  K7 · PELT STRUCTURAL BREAK DETECTION + PLOTLY REGIME CHART             ║
# ╚══════════════════════════════════════════════════════════════════════════╝
nat_ts = con.execute("""
    SELECT
        price_date,
        ROUND(AVG(price_close), 2) AS avg_close
    FROM engineered_features
    WHERE is_national_avg = 0
    GROUP BY price_date
    ORDER BY price_date
""").df()
nat_ts["price_date"] = pd.to_datetime(nat_ts["price_date"])

signal = nat_ts["avg_close"].values

# PELT with RBF cost — detects changes in mean AND variance
# pen=15 is well-calibrated for monthly price series; tune 5–30
algo = rpt.Pelt(model="rbf", min_size=6, jump=1).fit(signal)
bps  = algo.predict(pen=15)         # list of indices; last = len(signal)
bp_idx   = [b - 1 for b in bps[:-1]]   # convert to 0-indexed
bp_dates = nat_ts["price_date"].iloc[bp_idx].tolist()

print(f"Change-points detected: {len(bp_dates)}")
for i, d in enumerate(bp_dates):
    print(f"  Regime {i+1}→{i+2} boundary: {str(d)[:10]}  "
          f"(₦{signal[bp_idx[i]]:.0f}/L)")

# ── Assign regime labels to feat_df ──────────────────────────────────────────
def pelt_regime(d):
    r = 0
    for bp in bp_dates:
        if d >= bp: r += 1
    return r

feat_df["pelt_regime"] = feat_df["price_date"].apply(pelt_regime)
con.register("feat_df_v2", feat_df)
con.execute("CREATE OR REPLACE TABLE engineered_features AS SELECT * FROM feat_df_v2")

# ── Plotly figure ─────────────────────────────────────────────────────────────
REGIME_COLORS = [PAL["blue"], PAL["amber"], PAL["coral"], PAL["purple"], PAL["teal"]]

fig_pelt = go.Figure()

# Shade each regime segment
boundaries = [nat_ts["price_date"].iloc[0]] + bp_dates + [nat_ts["price_date"].iloc[-1]]
for i in range(len(boundaries)-1):
    fig_pelt.add_vrect(
        x0=boundaries[i], x1=boundaries[i+1],
        fillcolor=REGIME_COLORS[i % len(REGIME_COLORS)],
        opacity=0.10, line_width=0,
        annotation_text=f"<b>Regime {i}</b>",
        annotation_position="top left",
        annotation_font_size=10,
        annotation_font_color=REGIME_COLORS[i % len(REGIME_COLORS)],
    )

# Price series
fig_pelt.add_trace(go.Scatter(
    x=nat_ts["price_date"], y=nat_ts["avg_close"],
    mode="lines", name="National avg close",
    line=dict(color=PAL["teal"], width=2.2),
    hovertemplate="%{x|%b %Y}<br>₦%{y:,.2f}<extra></extra>",
))

# Vertical lines at break-points
for i, (d, idx) in enumerate(zip(bp_dates, bp_idx)):
    fig_pelt.add_vline(
        x=d.timestamp() * 1000,  # Plotly expects ms for datetime
        line_dash="dash",
        line_color=REGIME_COLORS[i % len(REGIME_COLORS)],
        line_width=2,
    )
    fig_pelt.add_annotation(
        x=d, y=signal[idx],
        text=f"BP{i+1}<br>{str(d)[:7]}",
        showarrow=True, arrowhead=2,
        arrowcolor=REGIME_COLORS[i % len(REGIME_COLORS)],
        font=dict(size=10, color=REGIME_COLORS[i % len(REGIME_COLORS)]),
        ax=30, ay=-40,
    )

apply_theme(
    fig_pelt,
    title="K7 · PELT Change-Point Detection — Nigeria Fuel Price Regimes",
    height=460,
    x_title="Date",
    y_title="Avg Close Price (₦/L)",
)
fig_pelt.update_yaxes(tickprefix="₦")
fig_pelt.show()

# Regime summary
reg_sum = feat_df.groupby("pelt_regime").agg(
    obs=("price_close","count"),
    mean_price=("price_close","mean"),
    std_price=("price_close","std"),
    from_date=("price_date","min"),
    to_date=("price_date","max"),
).round(2).reset_index()

fig_rs2 = go.Figure(go.Table(
    header=dict(
        values=[f"<b>{c}</b>" for c in reg_sum.columns],
        fill_color=PAL["coral"], font=dict(color="white",size=11),
        align="left", height=32,
    ),
    cells=dict(
        values=[reg_sum[c].tolist() for c in reg_sum.columns],
        align="left", height=28, font=dict(size=11),
        fill_color=[[PAL["light"],"white"]*4]*len(reg_sum.columns),
    )
))
apply_theme(fig_rs2, title="K7 · PELT Regime Summary Statistics", height=220)
fig_rs2.show()
```

## K8 — Insight 5: Multivariate LSTM with Lagged Exogenous Features
Build a 2-layer stacked LSTM trained exclusively on the post-subsidy regime with a 6-month sliding window.

```python
# ╔══════════════════════════════════════════════════════════════════════════╗
# ║  K8 · LSTM FEATURE PREPARATION — SQL + PANDAS                           ║
# ╚══════════════════════════════════════════════════════════════════════════╝
lstm_df = con.execute("""
    SELECT
        price_date,
        AVG(price_open)          AS price_open,
        AVG(price_high)          AS price_high,
        AVG(price_low)           AS price_low,
        AVG(price_close)         AS price_close,
        AVG(spread)              AS spread,
        AVG(momentum)            AS momentum,
        AVG(mom_pct)             AS mom_pct,
        AVG(roll_mean_3m)        AS roll_mean_3m,
        AVG(roll_std_12m)        AS roll_std_12m,
        AVG(trust_score)         AS trust_score,
        AVG(month_sin)           AS month_sin,
        AVG(month_cos)           AS month_cos,
        AVG(fuel_inflation_mom)  AS fuel_inflation_mom
    FROM engineered_features
    WHERE is_national_avg = 0
      AND pelt_regime >= 1       -- post-subsidy only
    GROUP BY price_date
    ORDER BY price_date
""").df()
lstm_df["price_date"] = pd.to_datetime(lstm_df["price_date"])
lstm_df = lstm_df.dropna().reset_index(drop=True)

# ── Synthetic exogenous series (replace with CBN / FRED data) ────────────────
n = len(lstm_df)
lstm_df["usd_ngn_rate"] = np.linspace(760, 1630, n) + np.random.normal(0, 35, n)
lstm_df["brent_crude"]  = np.linspace(82, 70, n)  + np.random.normal(0, 3, n)

FEATURE_COLS = [
    "price_open","price_high","price_low",
    "spread","momentum","mom_pct",
    "roll_mean_3m","roll_std_12m",
    "fuel_inflation_mom","trust_score",
    "month_sin","month_cos",
    "usd_ngn_rate","brent_crude",
]
TARGET_COL = "price_close"

feat_scaler   = MinMaxScaler()
target_scaler = MinMaxScaler()
X_sc = feat_scaler.fit_transform(lstm_df[FEATURE_COLS])
y_sc = target_scaler.fit_transform(lstm_df[[TARGET_COL]])

# ── Plotly: feature correlation heatmap ──────────────────────────────────────
corr = lstm_df[FEATURE_COLS + [TARGET_COL]].corr().round(3)
fig_corr = px.imshow(
    corr, text_auto=True, aspect="auto",
    color_continuous_scale="RdBu_r",
    color_continuous_midpoint=0,
    zmin=-1, zmax=1,
    title="K8 · Feature Correlation Matrix (post-subsidy regime)",
)
apply_theme(fig_corr, height=520)
fig_corr.update_traces(textfont_size=9)
fig_corr.show()

print(f"\n✅  LSTM input — {X_sc.shape[0]} samples × {X_sc.shape[1]} features")
```

```python
# ╔══════════════════════════════════════════════════════════════════════════╗
# ║  K8 · PYTORCH SLIDING-WINDOW DATASET + LSTM ARCHITECTURE                ║
# ╚══════════════════════════════════════════════════════════════════════════╝
WINDOW      = 6    # 6-month look-back (covers 2–4 week FX/crude lag)
TRAIN_SPLIT = 0.80

class FuelPriceDataset(Dataset):
    """Sliding-window multivariate dataset with trust weights."""
    def __init__(self, X: np.ndarray, y: np.ndarray,
                 trust: np.ndarray, window: int = 6):
        self.X, self.y, self.trust, self.window = (
            torch.FloatTensor(X), torch.FloatTensor(y),
            torch.FloatTensor(trust), window
        )

    def __len__(self):
        return len(self.X) - self.window

    def __getitem__(self, idx):
        return (self.X[idx:idx+self.window],
                self.y[idx+self.window],
                self.trust[idx+self.window])


class NigeriaFuelLSTM(nn.Module):
    """
    2-layer stacked LSTM for Nigeria retail fuel price forecasting.

    Architecture:
        Input  → LSTM (128 hidden, 2 layers, dropout 0.25)
               → FC(128 → 64) → ReLU
               → FC(64 → 1)
    """
    def __init__(self, n_feat: int, hidden: int = 128,
                 layers: int = 2, dropout: float = 0.25):
        super().__init__()
        self.lstm = nn.LSTM(
            input_size=n_feat, hidden_size=hidden, num_layers=layers,
            dropout=dropout if layers > 1 else 0.0, batch_first=True,
        )
        self.head = nn.Sequential(
            nn.Dropout(dropout),
            nn.Linear(hidden, 64), nn.ReLU(),
            nn.Linear(64, 1),
        )

    def forward(self, x):
        out, _ = self.lstm(x)          # (batch, window, hidden)
        return self.head(out[:, -1, :]) # last time-step → (batch, 1)


# ── Instantiate ───────────────────────────────────────────────────────────────
trust_sc = MinMaxScaler().fit_transform(
    lstm_df["trust_score"].values.reshape(-1,1)
).flatten()

n_train = int(len(X_sc) * TRAIN_SPLIT)
DEVICE  = torch.device("cuda" if torch.cuda.is_available() else "cpu")

train_ds = FuelPriceDataset(X_sc[:n_train], y_sc[:n_train], trust_sc[:n_train], WINDOW)
test_ds  = FuelPriceDataset(X_sc[n_train:], y_sc[n_train:], trust_sc[n_train:], WINDOW)
train_ld = DataLoader(train_ds, batch_size=16, shuffle=True,  drop_last=True)
test_ld  = DataLoader(test_ds,  batch_size=16, shuffle=False)

model = NigeriaFuelLSTM(n_feat=len(FEATURE_COLS)).to(DEVICE)

total_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(model)
print(f"\n✅  Device: {DEVICE}  |  Trainable params: {total_params:,}")
print(f"   Train samples: {len(train_ds)}  |  Test samples: {len(test_ds)}")
```

## K9 — Insight 6: Trust-Weighted Custom Loss + Training Loop
Replace standard MSE with a trust-weighted objective. High-confidence survey data penalises errors more heavily than spatially-interpolated LGA estimates.

```python
# ╔══════════════════════════════════════════════════════════════════════════╗
# ║  K9 · TRUST-WEIGHTED MSE LOSS — DEFINITION + PLOTLY VISUALISATION       ║
# ╚══════════════════════════════════════════════════════════════════════════╝
class TrustWeightedMSELoss(nn.Module):
    """
    Custom loss: L = mean( trust_norm × (ŷ − y)² )

    trust_norm = trust / (mean(trust) + ε)   [normalised to batch mean = 1]

    Effect:
        trust = 1.0  → verified NBS/WFP survey → penalise errors 100%
        trust = 0.3  → spatially interpolated  → penalise errors ~30%
        trust = 0.0  → unknown provenance      → no gradient signal

    The normalisation prevents total loss from collapsing when a batch
    contains uniformly low-trust observations.
    """
    def __init__(self, eps: float = 1e-8):
        super().__init__()
        self.eps = eps

    def forward(self, y_pred: torch.Tensor,
                y_true: torch.Tensor,
                trust:  torch.Tensor) -> torch.Tensor:
        sq_err  = (y_pred.squeeze() - y_true.squeeze()) ** 2
        t_norm  = trust / (trust.mean() + self.eps)
        return (t_norm * sq_err).mean()


# ── Plotly: loss surface visualisation ───────────────────────────────────────
trust_grid = np.linspace(0, 1, 120)
error_grid = np.linspace(0, 300, 120)  # ₦/L squared error range
T, E = np.meshgrid(trust_grid, error_grid)
Z = T * E

fig_surface = go.Figure(go.Surface(
    x=trust_grid, y=error_grid, z=Z,
    colorscale="Teal",
    opacity=0.85,
    hovertemplate="Trust: %{x:.2f}<br>Sq Error: %{y:.0f}<br>Loss: %{z:.2f}<extra></extra>",
))
fig_surface.update_layout(
    title=dict(text="K9 · Trust-Weighted Loss Surface — trust × squared_error",
               x=0.04, font=dict(size=15, color="#1A1A2E")),
    scene=dict(
        xaxis_title="Trust Score [0–1]",
        yaxis_title="Squared Prediction Error (₦²)",
        zaxis_title="Effective Loss",
        bgcolor="#F8F9FA",
    ),
    height=520,
    template="plotly_white",
    paper_bgcolor="white",
    font=dict(family="Inter, Arial, sans-serif", size=12),
    margin=dict(l=0, r=0, t=60, b=0),
)
fig_surface.show()

# ── 2-D slice: fixed squared error of ₦2500 ──────────────────────────────────
fixed_err = 2500
fig_slice = go.Figure()
fig_slice.add_trace(go.Scatter(
    x=trust_grid, y=trust_grid * fixed_err,
    mode="lines", name=f"Fixed sq-error = {fixed_err}",
    line=dict(color=PAL["teal"], width=2.5),
    fill="tozeroy", fillcolor="rgba(29,158,117,0.12)",
))
for tv, label in [(0.3, "Interpolated LGA"), (1.0, "Verified survey")]:
    fig_slice.add_vline(
        x=tv, line_dash="dash",
        line_color=PAL["coral"] if tv==1.0 else PAL["amber"],
        line_width=2,
        annotation_text=label, annotation_position="top",
        annotation_font_size=11,
    )
apply_theme(
    fig_slice,
    title="K9 · Trust Score vs Effective Loss Contribution (2-D Slice)",
    height=360,
    x_title="Trust Score",
    y_title="Effective Loss Contribution",
)
fig_slice.show()

print("✅  TrustWeightedMSELoss defined and visualised.")
```

```python
# ╔══════════════════════════════════════════════════════════════════════════╗
# ║  K9 · TRAINING LOOP — AdamW + ReduceLROnPlateau + Early Stopping        ║
# ╚══════════════════════════════════════════════════════════════════════════╝
EPOCHS   = 80
LR       = 1e-3
PATIENCE = 12

criterion = TrustWeightedMSELoss()
optimizer = torch.optim.AdamW(model.parameters(), lr=LR, weight_decay=1e-4)
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer, mode="min", factor=0.5, patience=6, verbose=False
)

train_losses, val_losses, lr_log = [], [], []
best_val, patience_count = float("inf"), 0

for epoch in range(1, EPOCHS+1):
    # ── Train ────────────────────────────────────────────────────────────────
    model.train()
    ep_train = 0.0
    for xb, yb, tb in train_ld:
        xb, yb, tb = xb.to(DEVICE), yb.to(DEVICE), tb.to(DEVICE)
        optimizer.zero_grad()
        loss = criterion(model(xb), yb, tb)
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        optimizer.step()
        ep_train += loss.item()
    avg_train = ep_train / len(train_ld)

    # ── Validate ─────────────────────────────────────────────────────────────
    model.eval()
    ep_val = 0.0
    with torch.no_grad():
        for xb, yb, tb in test_ld:
            xb, yb, tb = xb.to(DEVICE), yb.to(DEVICE), tb.to(DEVICE)
            ep_val += criterion(model(xb), yb, tb).item()
    avg_val = ep_val / max(len(test_ld), 1)

    train_losses.append(avg_train)
    val_losses.append(avg_val)
    lr_log.append(optimizer.param_groups[0]["lr"])
    scheduler.step(avg_val)

    if avg_val < best_val:
        best_val, patience_count = avg_val, 0
        torch.save(model.state_dict(), "best_nigeria_lstm.pt")
    else:
        patience_count += 1
        if patience_count >= PATIENCE:
            print(f"\n⏹  Early stopping at epoch {epoch}")
            break

    if epoch % 10 == 0 or epoch == 1:
        print(f"Epoch {epoch:3d}/{EPOCHS}  "
              f"Train: {avg_train:.6f}  Val: {avg_val:.6f}  "
              f"LR: {lr_log[-1]:.2e}")

print(f"\n✅  Training complete. Best val loss: {best_val:.6f}")

# ── Plotly loss + LR curve ────────────────────────────────────────────────────
epochs_ran = list(range(1, len(train_losses)+1))

fig_loss = make_subplots(
    rows=2, cols=1, shared_xaxes=True,
    row_heights=[0.70, 0.30],
    subplot_titles=("Training vs Validation Loss (Trust-Weighted MSE)",
                    "Learning Rate Schedule"),
    vertical_spacing=0.10,
)
fig_loss.add_trace(go.Scatter(
    x=epochs_ran, y=train_losses, name="Train loss",
    mode="lines", line=dict(color=PAL["blue"], width=2),
), row=1, col=1)
fig_loss.add_trace(go.Scatter(
    x=epochs_ran, y=val_losses, name="Val loss",
    mode="lines", line=dict(color=PAL["coral"], width=2),
), row=1, col=1)
fig_loss.add_trace(go.Scatter(
    x=epochs_ran, y=lr_log, name="LR",
    mode="lines", line=dict(color=PAL["amber"], width=1.5, dash="dot"),
    fill="tozeroy", fillcolor="rgba(244,162,38,0.12)",
), row=2, col=1)
fig_loss.update_layout(
    title=dict(text="K9 · LSTM Training Dynamics",
               x=0.04, font=dict(size=16, color="#1A1A2E")),
    height=520, template="plotly_white", paper_bgcolor="white",
    plot_bgcolor="#F8F9FA",
    font=dict(family="Inter, Arial, sans-serif", size=12),
    margin=dict(l=60, r=30, t=70, b=40),
    legend=dict(bgcolor="rgba(255,255,255,0.85)", bordercolor="#DDD", borderwidth=1),
)
fig_loss.update_xaxes(title_text="Epoch", row=2, col=1)
fig_loss.update_yaxes(showgrid=True, gridcolor="#ECECEC")
fig_loss.show()
```

## K10 — Model Evaluation & 3-Month Forward Forecast
Load best checkpoint, compute regression metrics, plot predicted vs actual, and generate the MSME early-warning forecast.

```python
# ╔══════════════════════════════════════════════════════════════════════════╗
# ║  K10 · EVALUATION — METRICS + PREDICTED vs ACTUAL + RESIDUALS           ║
# ╚══════════════════════════════════════════════════════════════════════════╝
model.load_state_dict(torch.load("best_nigeria_lstm.pt", map_location=DEVICE))
model.eval()

all_preds, all_truths = [], []
with torch.no_grad():
    for xb, yb, _ in test_ld:
        preds = model(xb.to(DEVICE)).cpu().numpy()
        all_preds.extend(preds.flatten())
        all_truths.extend(yb.numpy().flatten())

y_pred_inv = target_scaler.inverse_transform(
    np.array(all_preds).reshape(-1,1)).flatten()
y_true_inv = target_scaler.inverse_transform(
    np.array(all_truths).reshape(-1,1)).flatten()
residuals  = y_true_inv - y_pred_inv

mae  = mean_absolute_error(y_true_inv, y_pred_inv)
rmse = np.sqrt(mean_squared_error(y_true_inv, y_pred_inv))
r2   = r2_score(y_true_inv, y_pred_inv)
mape = np.mean(np.abs(residuals / (y_true_inv + 1e-8))) * 100

# ── Metrics table ─────────────────────────────────────────────────────────────
metrics = pd.DataFrame({
    "Metric"      : ["MAE (₦/L)","RMSE (₦/L)","MAPE (%)","R²"],
    "Value"       : [f"₦{mae:.2f}", f"₦{rmse:.2f}",
                     f"{mape:.2f}%", f"{r2:.4f}"],
    "Interpretation": [
        "Mean absolute price error",
        "Penalises large outlier errors",
        "% error relative to actual price",
        "Explained variance (1.0 = perfect)",
    ]
})
fig_metrics = go.Figure(go.Table(
    header=dict(
        values=["<b>Metric</b>","<b>Value</b>","<b>Interpretation</b>"],
        fill_color=PAL["teal"], font=dict(color="white",size=12),
        align="left", height=32,
    ),
    cells=dict(
        values=[metrics[c].tolist() for c in metrics.columns],
        align="left", height=30,
        font=dict(size=12),
        fill_color=[[PAL["light"],"white",PAL["light"],"white"]]*3,
    )
))
apply_theme(fig_metrics, title="K10 · Test Set Evaluation Metrics", height=240)
fig_metrics.show()

# ── Predicted vs Actual time-series ──────────────────────────────────────────
test_dates = lstm_df["price_date"].iloc[
    n_train + WINDOW : n_train + WINDOW + len(y_pred_inv)
].values

fig_pred = make_subplots(
    rows=2, cols=1,
    shared_xaxes=True,
    row_heights=[0.65, 0.35],
    subplot_titles=(
        "Predicted vs Actual Close Price (₦/L) — Test Set",
        "Residuals (Actual − Predicted)",
    ),
    vertical_spacing=0.08,
)
fig_pred.add_trace(go.Scatter(
    x=test_dates, y=y_true_inv,
    mode="lines", name="Actual",
    line=dict(color=PAL["teal"], width=2.2),
    hovertemplate="%{x|%b %Y}<br>Actual: ₦%{y:,.2f}<extra></extra>",
), row=1, col=1)
fig_pred.add_trace(go.Scatter(
    x=test_dates, y=y_pred_inv,
    mode="lines", name="LSTM Forecast",
    line=dict(color=PAL["coral"], width=2, dash="dash"),
    hovertemplate="%{x|%b %Y}<br>Forecast: ₦%{y:,.2f}<extra></extra>",
), row=1, col=1)

# Confidence band (±1 rolling std of residuals)
roll_std = pd.Series(residuals).rolling(3, min_periods=1).std().fillna(0).values
fig_pred.add_trace(go.Scatter(
    x=np.concatenate([test_dates, test_dates[::-1]]),
    y=np.concatenate([y_pred_inv + roll_std, (y_pred_inv - roll_std)[::-1]]),
    fill="toself", fillcolor="rgba(216,90,48,0.10)",
    line=dict(color="rgba(255,255,255,0)"),
    hoverinfo="skip", name="±1σ band",
), row=1, col=1)

# Residuals
fig_pred.add_trace(go.Bar(
    x=test_dates, y=residuals,
    marker_color=[PAL["teal"] if r>=0 else PAL["coral"] for r in residuals],
    name="Residual",
    hovertemplate="%{x|%b %Y}<br>Residual: ₦%{y:,.2f}<extra></extra>",
), row=2, col=1)
fig_pred.add_hline(y=0, line_dash="dot", line_color=PAL["slate"],
                   line_width=1, row=2, col=1)

fig_pred.update_layout(
    title=dict(text="K10 · Model Evaluation — Predicted vs Actual",
               x=0.04, font=dict(size=16, color="#1A1A2E")),
    height=620, template="plotly_white", paper_bgcolor="white",
    plot_bgcolor="#F8F9FA",
    font=dict(family="Inter, Arial, sans-serif", size=12),
    margin=dict(l=60, r=30, t=70, b=40),
    legend=dict(bgcolor="rgba(255,255,255,0.85)", bordercolor="#DDD", borderwidth=1),
)
fig_pred.update_yaxes(tickprefix="₦", showgrid=True, gridcolor="#ECECEC")
fig_pred.show()

# ── Residual distribution ─────────────────────────────────────────────────────
fig_hist = go.Figure()
fig_hist.add_trace(go.Histogram(
    x=residuals, nbinsx=30,
    marker_color=PAL["blue"], opacity=0.80, name="Residuals",
    hovertemplate="Residual: ₦%{x:,.1f}<br>Count: %{y}<extra></extra>",
))
fig_hist.add_vline(x=0, line_dash="dash",
                   line_color=PAL["coral"], line_width=2,
                   annotation_text="Zero error",
                   annotation_position="top")
fig_hist.add_vline(x=residuals.mean(), line_dash="dot",
                   line_color=PAL["amber"], line_width=1.8,
                   annotation_text=f"Mean: ₦{residuals.mean():.1f}",
                   annotation_position="top right")
apply_theme(
    fig_hist,
    title="K10 · Residual Distribution (should be zero-centred & symmetric)",
    height=360, x_title="Residual (₦/L)", y_title="Frequency",
)
fig_hist.show()
```

```python
# ╔══════════════════════════════════════════════════════════════════════════╗
# ║  K10 · 3-MONTH AUTOREGRESSIVE FORWARD FORECAST (MSME EARLY WARNING)     ║
# ╚══════════════════════════════════════════════════════════════════════════╝
model.eval()
HORIZON = 3
current_w = torch.FloatTensor(X_sc[-WINDOW:]).unsqueeze(0).to(DEVICE)

fc_prices = []
with torch.no_grad():
    for _ in range(HORIZON):
        pred_sc  = model(current_w).cpu().numpy()
        pred_ngn = target_scaler.inverse_transform(pred_sc.reshape(-1,1))[0,0]
        fc_prices.append(round(float(pred_ngn), 2))

        new_step    = current_w[0, -1, :].clone().cpu().numpy()
        new_step[3] = pred_sc[0, 0]   # update price_close slot
        current_w   = torch.cat(
            [current_w[:, 1:, :],
             torch.FloatTensor(new_step).unsqueeze(0).unsqueeze(0).to(DEVICE)],
            dim=1,
        )

last_date = lstm_df["price_date"].iloc[-1]
fc_dates  = pd.date_range(start=last_date + pd.DateOffset(months=1),
                           periods=HORIZON, freq="MS")
fc_df     = pd.DataFrame({"month": fc_dates, "forecast_ngn": fc_prices})

last_actual = float(target_scaler.inverse_transform(y_sc[-1:]).flatten()[0])
fc_df["vs_current_pct"] = ((fc_df["forecast_ngn"] - last_actual) / last_actual * 100).round(2)

# ── Plotly combined historical + forecast chart ───────────────────────────────
hist_line = lstm_df[["price_date","price_close"]].tail(24)

fig_fc = go.Figure()

fig_fc.add_trace(go.Scatter(
    x=hist_line["price_date"], y=hist_line["price_close"],
    mode="lines", name="Historical (national avg)",
    line=dict(color=PAL["teal"], width=2.2),
    hovertemplate="%{x|%b %Y}<br>₦%{y:,.2f}<extra></extra>",
))

# Connect historical to forecast with a bridge
bridge_x = [hist_line["price_date"].iloc[-1], fc_dates[0]]
bridge_y = [hist_line["price_close"].iloc[-1], fc_prices[0]]
fig_fc.add_trace(go.Scatter(
    x=bridge_x, y=bridge_y,
    mode="lines", line=dict(color=PAL["amber"], width=1.5, dash="dot"),
    showlegend=False, hoverinfo="skip",
))

fig_fc.add_trace(go.Scatter(
    x=fc_dates, y=fc_prices,
    mode="lines+markers", name="3-Month Forecast",
    line=dict(color=PAL["coral"], width=2.5, dash="dash"),
    marker=dict(size=12, color=PAL["coral"], symbol="diamond",
                line=dict(color="white", width=2)),
    hovertemplate="%{x|%b %Y}<br><b>Forecast: ₦%{y:,.2f}</b><extra></extra>",
))

# Annotation labels on forecast points
for d, p, pct in zip(fc_dates, fc_prices, fc_df["vs_current_pct"]):
    color = PAL["coral"] if pct > 5 else (PAL["teal"] if pct < -3 else PAL["amber"])
    arrow = "📈" if pct > 5 else ("📉" if pct < -3 else "➡️")
    fig_fc.add_annotation(
        x=d, y=p,
        text=f"<b>₦{p:,.0f}</b><br>{arrow} {pct:+.1f}%",
        showarrow=True, arrowhead=2, arrowcolor=color,
        font=dict(size=11, color=color),
        bgcolor="white", bordercolor=color, borderwidth=1,
        ax=0, ay=-55,
    )

fig_fc.add_vrect(
    x0=str(fc_dates[0].date()), x1=str(fc_dates[-1].date()),
    fillcolor=PAL["coral"], opacity=0.05,
    line_width=0,
    annotation_text="Forecast window",
    annotation_position="top left",
    annotation_font_size=10,
)

apply_theme(
    fig_fc,
    title="K10 · MSME Early Warning System — 3-Month Fuel Price Forecast",
    height=480,
    x_title="Date",
    y_title="Retail Price (₦/L)",
)
fig_fc.update_yaxes(tickprefix="₦")
fig_fc.show()

# ── Console summary ───────────────────────────────────────────────────────────
print("\n" + "═"*52)
print("  🔮  3-MONTH FORWARD FORECAST — EARLY WARNING SYSTEM")
print("═"*52)
for _, row in fc_df.iterrows():
    m   = row["month"].strftime("%b %Y")
    p   = row["forecast_ngn"]
    pct = row["vs_current_pct"]
    sig = "⚠️  SPIKE WARNING" if pct>5 else ("✅ STABLE" if abs(pct)<=5 else "📉 RELIEF")
    print(f"  {m}  →  ₦{p:>8,.2f}/L   ({pct:+.1f}%)  {sig}")
print("═"*52)

# ── CSV exports ───────────────────────────────────────────────────────────────
fc_df.to_csv("nigeria_fuel_3m_forecast.csv", index=False)
feat_df.to_csv("nigeria_fuel_engineered_features.csv", index=False)
print("\n✅  Exports: nigeria_fuel_3m_forecast.csv, nigeria_fuel_engineered_features.csv")
```

### Suggested Methodology Chapter Wording
> *"We employ a two-layer stacked LSTM (Hochreiter & Schmidhuber, 1997) trained exclusively on the post-subsidy regime as objectively identified by the Pruned Exact Linear Time (PELT) algorithm (Killick et al., 2012). The model minimises a trust-weighted MSE objective in which each observation's gradient contribution is proportional to its WFP data quality score, suppressing noise from spatially interpolated LGA estimates. A six-month sliding window of 14 engineered features — including OHLC derivatives, cyclical month encodings, and lagged exogenous FX and crude proxies — feeds the model to produce a three-month early-warning price forecast."*
